```python
# load in csv data
df = pd.read_csv('/colab_notebooks/MyDrive/Colab Notebooks/netflix_titles.csv')
```

```python
# check column data types
print(df.dtypes)
```

```python
# check for dup ids, row count was 8807
print(df['show_id'].nunique())
```

```python
# check for blanks and white spaces
df.select_dtypes(include='object').apply(lambda col: col.astype(str).str.strip().eq('').sum())
```

```python
# check for nulls, nans
print(df.isnull().sum())
```

```python
# check for something interesting.
# What is the most common rating, type of media like show or movie, and
# what is the average duration of the most popular common media?
print(df['rating'].value_counts())
```

```python
# finding the most common media type
print(df['type'].value_counts())
```

```python
# percentage make up between movies and tv shows to the total
print(df['type'].value_counts(normalize=True) * 100)
```

```python
# get average duration of for movies. First extracting text to string and then averaging.
# Average movie duration (in minutes)
print(df[df['type']=='Movie']['duration'].str.replace(' min','',regex=False).astype(float).mean())
```

```python
# convert minutes to hours and minutes
avg_duration = df[df['type']=='Movie']['duration'].str.replace(' min','',regex=False).astype(float).mean()

hours, minutes = divmod(round(avg_duration), 60)
print(f"{hours}h {minutes}m")
```
